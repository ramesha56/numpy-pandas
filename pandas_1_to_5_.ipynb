```python
import pandas as pd
```

```python
##task 1


data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)

print(df)

print("avrage :", df["marks"].mean())
print("maximum :", df["marks"].max())
print("minimum :", df["marks"].min())
print("total :", len(df))
```

```python
###task 2

print(df[df["marks"] >= 50])
```

```python
##task 4

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Course": ["AI", "Web", "AI", "Web", "AI"],
    "marks": [70, 85, 60, 95, 45]
}

df2 = pd.DataFrame(data)

print(df2.groupby("Course")["marks"].mean())
```
